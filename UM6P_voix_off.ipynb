```python
import asyncio
import edge_tts
import nest_asyncio
from IPython.display import Audio

# Apply nest_asyncio to allow async execution within Colab
nest_asyncio.apply()

# TEXT CORRECTION:
# I replaced "6" with "sisse" to force the pronunciation /sis/ instead of /si/.
text = """
Au cœur du Maroc, à Ben Guerir, l’Université Mohammed Sisse Polytechnique, l’U M Sisse P, est une université de nouvelle génération, pensée pour répondre aux grands défis de notre temps.
Sa mission est claire : mettre la science, l’innovation et l’entrepreneuriat au service du développement durable, avec une ambition affirmée pour l’Afrique.

À l’U M Sisse P, la recherche ne reste pas dans les laboratoires : elle se transforme en solutions concrètes. L’université s’appuie sur un écosystème unique, qui réunit des chercheurs, des étudiants, des ingénieurs, des entrepreneurs et des partenaires publics et privés. Ensemble, ils travaillent sur des enjeux majeurs : la transition énergétique, l’eau, les matériaux, le numérique, la santé, et bien sûr, l’agriculture et l’environnement.

Dans ce paysage, le College of Agriculture and Environmental Sciences, le C A E S, occupe une place centrale. Parce que l’agriculture est au carrefour de tout : sécurité alimentaire, climat, biodiversité, eau, emplois et territoires. Et parce que les systèmes agroalimentaires doivent évoluer rapidement, avec des réponses à la fois scientifiques, économiques et sociales.

Le C A E S porte une vision : concevoir des trajectoires de transition réalistes et ambitieuses, adaptées aux contextes marocains et africains. Nos équipes travaillent sur la santé des sols, la nutrition des plantes, la gestion de l’eau et l’irrigation, l’agroécologie, la diversité au champ et les relations symbiotiques entre plantes et insectes, la résilience climatique, la restauration des écosystèmes, et la valorisation durable de la biomasse.

Nous mobilisons également la data science, le digital et l’observation de la Terre pour mieux comprendre les dynamiques des territoires, suivre les cultures, anticiper les risques et appuyer la décision.

Former les talents est au cœur de notre action. Le C A E S développe des programmes d’excellence, du pré-doctorat au doctorat, et des formations professionnalisantes, en lien étroit avec les besoins du terrain. Notre objectif : former une nouvelle génération capable de relier science, innovation et impact.

Enfin, le C A E S s’inscrit dans une dynamique de partenariats internationaux et de collaboration avec les acteurs du développement, pour accélérer le passage à l’échelle : du laboratoire au champ, du prototype à la politique publique.

L’U M Sisse P et le C A E S partagent une conviction : face aux défis agricoles et environnementaux, nous avons besoin de connaissances solides, d’alliances fortes et d’actions rapides.
C’est cette énergie collective que nous construisons, chaque jour, à Ben Guerir, pour le Maroc, et pour l’Afrique.
"""

# Using the high quality French voice
VOICE = "fr-FR-VivienneNeural"
OUTPUT_FILE = "voix_off_um6p_corrected.mp3"

async def generate_voice():
    communicate = edge_tts.Communicate(text, VOICE)
    await communicate.save(OUTPUT_FILE)

# Run the generation
loop = asyncio.get_event_loop()
loop.run_until_complete(generate_voice())

print(f"File saved as: {OUTPUT_FILE}")
Audio(OUTPUT_FILE)
```
